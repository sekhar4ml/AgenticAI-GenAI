```python
from langchain.chains import ConversationChain
from langchain.memory import ConversationBufferMemory
```

```python
memory=ConversationBufferMemory()
```

```python
from langchain_ollama.llms import OllamaLLM

inf_llm=OllamaLLM(model="chevalblanc/gpt-4o-mini")
```

```python
conversation=ConversationChain(llm=inf_llm,memory=memory)
```

```python
conversation.predict(input="what is the capital city of OHIO.Answer the response in English")
```

```python
conversation.predict(input="What are the top 5 attraction in it")
```

```python
conversation.predict(input="What are child attractions there")
```
