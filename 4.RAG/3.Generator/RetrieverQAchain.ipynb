```python
from langchain_community.document_loaders import Docx2txtLoader
```

```python
loader=Docx2txtLoader("/Users/sreesekhar/IAI/AgenticGenerativeAI/week9/lab/data/fitness.docx")

docs=loader.load()
```

```python
from langchain_huggingface.embeddings import HuggingFaceEmbeddings
```

```python
embedding_llm=HuggingFaceEmbeddings(model_name="sentence-transformers/all-mpnet-base-v2")
```

```python
from langchain_community.vectorstores import FAISS
```

```python
doc_db=FAISS.from_documents(docs,embedding_llm)
```

```python
retriever=doc_db.as_retriever()
```

```python
from langchain_ollama.llms import OllamaLLM

inf_llm=OllamaLLM(model="chevalblanc/gpt-4o-mini")
```

```python
from langchain.chains import RetrievalQA
```

```python
qa_chain=RetrievalQA.from_chain_type(
    
    llm=inf_llm,
    chain_type="stuff",
    retriever=retriever
)
```

```python
qa_chain.invoke("what are the fitness center ethics")
```

```python
from langchain.chains import RetrievalQAWithSourcesChain

qa_chain=RetrievalQAWithSourcesChain.from_chain_type(
    
    llm=inf_llm,
    chain_type="stuff",
    retriever=retriever,
    return_source_documents=True
)
```

```python
qa_chain.invoke("what are the fitness center ethics")
```
