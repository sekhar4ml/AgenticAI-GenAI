```python
from langchain.chains import LLMChain
from langchain.prompts import PromptTemplate,ChatPromptTemplate
```

```python
from langchain_ollama.llms import OllamaLLM

chat_model=OllamaLLM(model="chevalblanc/gpt-4o-mini")
```

```python
template="what is the capitol city{state}.language is english"
```

```python
prompt=PromptTemplate(template=template,input_variables=["state"])
```

```python
chain=LLMChain(llm=chat_model,prompt=prompt)
```

```python
chain('ohio')
```

```python
chain('Texas')
```
