```python
from langchain_community.document_loaders import PyPDFLoader

loader=PyPDFLoader("/Users/sreesekhar/IAI/AgenticGenerativeAI/week9/lab/data/chase_banking.pdf")
documents=loader.load()

print(documents)
```

```python
from langchain_community.vectorstores import FAISS
from langchain_huggingface.embeddings import  HuggingFaceEmbeddings

embedding_llm=HuggingFaceEmbeddings(model_name="sentence-transformers/all-mpnet-base-v2")
```

```python
faiss_db=FAISS.from_documents(documents,embedding_llm)
```

```python
dbtest=faiss_db.similarity_search("how to open checking account")
dbtest
```

```python
retriever=faiss_db.as_retriever()
```

```python
from langchain_ollama.llms import OllamaLLM

inf_llm=OllamaLLM(model="chevalblanc/gpt-4o-mini")
```

```python
from langchain.memory import ConversationBufferMemory

memory_buffer=ConversationBufferMemory(
    
    memory_key="chat_history",
    return_messages=True
)
```

```python
from langchain.chains import ConversationalRetrievalChain

cnvr_chain=ConversationalRetrievalChain.from_llm(
    
    llm=inf_llm,
    retriever=retriever,
    memory=memory_buffer
)
```

```python
cnvr_chain.invoke("how to open a checking account")
```

```python
cnvr_chain.invoke("how to avoid over draft fees. Answer in English")
```
