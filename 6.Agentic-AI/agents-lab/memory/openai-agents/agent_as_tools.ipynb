```python
from agents import Agent,Runner
```

```python
spanish_agent = Agent(
    
    name="Spanish Translator",
    instructions="You are a helpful assistant that translates English text to Spanish.",
)

french_agent = Agent(
    
    name="French Translator",
    instructions="You are a helpful assistant that translates English text to French.",
)
```

```python
tools=[
    
    spanish_agent.as_tool(
        tool_name="translate_to_spanish",
        tool_description="Translates English text to Spanish.",
        
        ),
    
    french_agent.as_tool(
        
        tool_name="translate_to_french",
        tool_description="Translates English text to French.",
    )
    
]
```

```python
translator_agent=Agent(
    
    name="Multi-language Translator",
    instructions="You are a helpful assistant that translates English text to multiple languages as requested by user",
    
    tools=tools
    
)
```

```python
response= await Runner.run(translator_agent, "Translate 'Good morning' to Spanish and French.")

print(response)
```
