```python
from agents import Agent, Runner, function_tool
from dotenv import load_dotenv
load_dotenv()
```

```python
@function_tool
def add_numbers(a: int, b: int) -> int:
    """Add two numbers and return the result."""
    return a + b
```

```python
tools=[add_numbers]
```

```python
agent=Agent(
    
    name="Addition Agent",
    instructions="You are a caliculator agent. Use tools when helpful.",
    tools=tools,
)
```

```python
response = await Runner.run(agent, "What is the sum of 15 and 27?")
print(response)
```
