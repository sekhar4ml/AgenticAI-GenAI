```python
from agents import Agent, Runner
from agents import WebSearchTool
```

```python
budget_estimator = Agent(
    
    name="Trip Budget Estimator",
    instructions="You estimate the budget for a trip based on user inputs such as destination, duration, accommodation type, and activities planned.You can search the web for current prices if needed.",
    tools=[WebSearchTool()],
    handoff_description="Travel cost estimation expert.",
    model="gpt-4o"
)
```

```python
iternary_agent = Agent(
    
    name="Trip Itinerary Planner",
    instructions="You create a detailed trip itinerary based on user preferences such as destination, duration, interests, and budget. You can search the web for attractions, restaurants, and events if needed.",
    tools=[WebSearchTool()],
    handoff_description="Travel itinerary planning expert.",
    model="gpt-4o"
)
```

```python
flight_hotel_agent = Agent(
    name="Flight and Hotel Booker",
    instructions="You assist users in finding and booking flights and hotels based on their trip details such as destination, dates, budget, and preferences. You can search the web for current flight and hotel options if needed.",
    handoff_description="Flight and Hotel booking expert.",
    tools=[WebSearchTool()],
    model="gpt-4o"
    
)
```

```python
local_guide_agent = Agent(
    
    name="Local Guide",
    instructions="You provide local recommendations and tips for travelers based on their destination. This includes restaurants, attractions, cultural norms, and safety advice. You can search the web for current information if needed.",
    tools=[WebSearchTool()],
    handoff_description="Local travel guide expert.",
    model="gpt-4o"
)
```

```python
master_travel_agent = Agent(
    
    name="Master Travel Agent",
    instructions="You coordinate the efforts of specialized travel agents to provide comprehensive travel planning services. You delegate tasks such as budget estimation, itinerary planning, flight and hotel booking, and local guidance to the appropriate agents based on user requests.",
    model="gpt-4o",
    handoffs=[local_guide_agent, flight_hotel_agent, iternary_agent, budget_estimator]
    
)
```

```python
question="I want to travel to India "
```

```python
response=await Runner.run(master_travel_agent, question)

print(response.final_output)
```
