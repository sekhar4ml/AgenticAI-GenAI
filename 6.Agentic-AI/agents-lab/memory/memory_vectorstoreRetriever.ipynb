```python
import faiss

from langchain_community.docstore import InMemoryDocstore
from langchain_community.vectorstores import FAISS
from langchain_openai import OpenAIEmbeddings
```

```python
embedding_size=1536

index = faiss.IndexFlatL2(embedding_size)
embedding_fn = OpenAIEmbeddings().embed_query

vectorstore = FAISS(embedding_function=embedding_fn, index=index, docstore=InMemoryDocstore({}), index_to_docstore_id={})

```

```python
from langchain_openai import ChatOpenAI

llm = ChatOpenAI(model_name="gpt-4o", temperature=0)
```

```python
from langchain_classic.memory import VectorStoreRetrieverMemory
from langchain_classic.chains import ConversationChain

memory=VectorStoreRetrieverMemory(vectorstore=vectorstore,
                                  
                                  retriever=vectorstore.as_retriever(),
                                  llm=llm,memory_key="chat_history",)     
```

```python
from langchain_classic.agents import load_tools

tools=load_tools(["wikipedia","llm-math"], llm=llm)
```

```python
from langchain_classic.agents import initialize_agent, AgentType

agent=initialize_agent(
    
    tools=tools,
    llm=llm,
    memory=memory,
    
    agent=AgentType.CONVERSATIONAL_REACT_DESCRIPTION,
    verbose=True,
    max_iterations=5,
    handle_parsing_errors=True
)
```

```python
agent.run(input="What is the square root of the number of people on wikipedia as of 2024-06-01?")
```
