```python
from langchain_classic.agents import initialize_agent, load_tools
from langchain_classic.agents import AgentType

from langchain_classic.llms import Ollama
```

```python
llm=Ollama(model="llama3.2")
```

```python
import yfinance as yf
from langchain_classic.agents import tool
```

```python
@tool

def get_stock_price(ticker: str) -> str:
    """Get the current stock price for a given ticker symbol."""
    stock = yf.Ticker(ticker)
    price = stock.info.get('regularMarketPrice', 'N/A')
    return f"The current stock price of {ticker} is {price}."
```

```python
@tool

def get_time(text:str) -> str:
    """Get the current time."""
    from datetime import datetime
    return str(datetime.now())
```

```python
tools=[get_stock_price, get_time]
```

```python
agent=initialize_agent(
    tools=tools, 
    llm=llm, 
    agent=AgentType.CHAT_ZERO_SHOT_REACT_DESCRIPTION, 
    handle_parsing_errors=True,
    verbose=True)
```

```python
print(agent.run("What is the current stock price of AAPL and what time is it now?"))
```
