```python
from langchain_classic.agents import initialize_agent, AgentType
```

```python
from langchain_classic.llms import Ollama
llm=Ollama(model="llama3.2")
```

```python
from openai import OpenAI
```

```python
prompt=("You are an expert in postgresql databases. i will ask you questions about postgresql and you will answer it. If you don't know the answer, just say you don't know. Do not try to make up an answer.")
```

```python
from langchain_community.tools.sql_database.tool import QuerySQLDataBaseTool,SQLDatabase
```

```python
db = SQLDatabase.from_uri(
    "postgresql+psycopg://postgres:pg1234@localhost:5432/srg"
)
```

```python
db_tool = QuerySQLDataBaseTool(db=db, llm=llm, prompt=prompt)
```

```python
tools = [db_tool]
```

```python
agent = initialize_agent(
    
    tools=[db_tool],
    llm=llm,
    agent=AgentType.ZERO_SHOT_REACT_DESCRIPTION,
    max_iterations=10,
    handle_parsing_errors=True,
    verbose=True,
)
```

```python
agent.run("how many tables are there in the databse?")
```

```python
agent.run("list all the tables in the database")
```

```python
agent.run("What are the different types of jobs with salary range between 50000 to 100000?")
```
