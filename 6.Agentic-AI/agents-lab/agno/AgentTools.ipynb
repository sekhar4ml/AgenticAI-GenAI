```python
from agno.agent import Agent
from agno.models.ollama import Ollama

from agno.tools.reasoning  import ReasoningTools
from agno.tools.yfinance import YFinanceTools
```

```python
myAgent=Agent(
    
    name="FinanceAgent",
    model=Ollama(id="llama3.2"),
    
    tools=[
        
        ReasoningTools(add_instructions=True),
        YFinanceTools()
    ],
    
    instructions=[
        
        "You are a financial data analysis agent.",
        "Use table to display data when appropriate.",
        "only output the report ."
    ],
    markdown=True
    
)
```

```python
response=myAgent.run(
    
    "Create a report on AAPL and MSFT including current price."
    "basic company info, add analyst recommendations."
)

print(response.content)
```
