```python
from agno.agent import Agent

from agno.models.ollama import Ollama
```

```python
agent=Agent(
    
    name="Simple Agent",
    model=Ollama(id="llama3.2"),
    description="An example agent using Ollama's Llama 3.2 model.",
    markdown=True
)
```

```python
response=agent.run("Explain the theory of relativity in simple terms.")

print(response.content)
```
