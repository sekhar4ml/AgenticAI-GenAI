```python
from langchain_community.document_loaders import PyPDFLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter

from langchain_openai import OpenAIEmbeddings

from langchain_community.vectorstores import FAISS

```

```python
loader=PyPDFLoader("/Users/sreesekhar/IAI/1.AgenticAI/week-8/agents-lab/agno/nke-10k-2023.pdf")

docs=loader.load()

text_splitter = RecursiveCharacterTextSplitter(chunk_size=1000, chunk_overlap=200)

chunks=text_splitter.split_documents(docs)

embeddings=OpenAIEmbeddings()

faiss_db=FAISS.from_documents(chunks, embeddings)

retriver=faiss_db.as_retriever()

```

```python
from agno.knowledge.knowledge import Knowledge
from agno.vectordb.langchaindb import LangChainVectorDb

rag_knowledge=Knowledge(vector_db=LangChainVectorDb(knowledge_retriever=retriver))
```

```python
from agno.agent import Agent
from agno.models.ollama import Ollama
```

```python
agent=Agent(
    
    name="Simple Agent",
    model=Ollama(id="llama3.2"),
    knowledge=rag_knowledge,
    search_knowledge=True,
    description="Answer questions from Knowlddge base only",
    markdown=True
)
```

```python
response=agent.run(" what is the business model of nike as per their 10k filing?")

print(response.content)
```
